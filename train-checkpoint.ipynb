```python
# train.py

import os
import torch
from torch import nn, optim
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader, random_split
import matplotlib.pyplot as plt
from utils import EarlyStopping, calculate_accuracy, plot_confusion_matrix, GradCAM

# Check device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Directories
train_dir = 'Brain-Tumor-Classification-DataSet-master/Training'
test_dir = 'Brain-Tumor-Classification-DataSet-master/Testing'

# Data Transforms
train_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(),
    transforms.ToTensor(),
    transforms.Normalize([0.5], [0.5])
])

test_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.5], [0.5])
])

# Load datasets
full_train_dataset = datasets.ImageFolder(train_dir, transform=train_transforms)
test_dataset = datasets.ImageFolder(test_dir, transform=test_transforms)

# Split train into train/val
train_size = int(0.8 * len(full_train_dataset))
val_size = len(full_train_dataset) - train_size
train_dataset, val_dataset = random_split(full_train_dataset, [train_size, val_size])

# Dataloaders
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

# Model
model = models.resnet18(pretrained=True)
model.fc = nn.Linear(model.fc.in_features, 4)  # 4 classes
model = model.to(device)

# Loss and Optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

# EarlyStopping
early_stopping = EarlyStopping(patience=5, verbose=True, save_path="models/best_model.pth")

# Training Loop
epochs = 50
train_losses, val_losses = [], []

for epoch in range(epochs):
    model.train()
    running_loss = 0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    train_losses.append(running_loss/len(train_loader))

    # Validation
    model.eval()
    val_loss = 0
    y_true, y_pred = [], []
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            val_loss += loss.item()

            preds = outputs.argmax(dim=1)
            y_true.append(labels.cpu())
            y_pred.append(preds.cpu())

    val_losses.append(val_loss/len(val_loader))

    # EarlyStopping
    early_stopping(val_loss/len(val_loader), model)
    if early_stopping.early_stop:
        print("Early stopping triggered")
        break

    print(f"Epoch {epoch+1}/{epochs}.. Train Loss: {running_loss/len(train_loader):.3f}.. Val Loss: {val_loss/len(val_loader):.3f}")

# Load Best Model
model.load_state_dict(torch.load("models/best_model.pth"))

# Evaluation
model.eval()
y_true, y_pred = [], []
with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        preds = outputs.argmax(dim=1)
        y_true.extend(labels.cpu().numpy())
        y_pred.extend(preds.cpu().numpy())

accuracy = calculate_accuracy(torch.tensor(y_true), torch.tensor(y_pred))
print(f"Test Accuracy: {accuracy*100:.2f}%")
plot_confusion_matrix(y_true, y_pred, classes=test_dataset.classes)

# GradCAM Example
gradcam = GradCAM(model, model.layer4[1].conv2)  # Choose a deep layer

sample_img, _ = next(iter(test_loader))
sample_img = sample_img[0].unsqueeze(0).to(device)
heatmap = gradcam.generate_heatmap(sample_img)

# Plot GradCAM
plt.imshow(sample_img.squeeze().permute(1,2,0).cpu().numpy()*0.5 + 0.5)
plt.imshow(heatmap, cmap='jet', alpha=0.5)
plt.title("GradCAM Visualization")
plt.axis('off')
plt.savefig('outputs/gradcam_sample.png')
plt.show()
```
